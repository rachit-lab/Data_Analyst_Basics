```python
import numpy as np

arr1 = np.random.randint(1,21,size=(5,5))
print("Original Array :", arr1)

arr_mean = np.mean(arr1)
arr_median = np.median(arr1)
arr_std = np.std(arr1)
arr_var = np.var(arr1)

print(arr_mean)
print(arr_median)
print(arr_std)
print(arr_var)
```

```python
arr1 = np.random.randint(1,9,size=(4,4))
print("Original Array : \n",arr1)

mean = np.mean(arr1)
std = np.std(arr1)

normalised_arr = (arr1 - mean) / std

print(normalised_arr)
```

```python
matrix = np.random.randint(1,11,size=(3,3))
print("Original Matrix : \n", matrix)

det = np.linalg.det(matrix)
inverse = np.linalg.inv(matrix)
equilv = np.linalg.eigvals(matrix)

print(det)
print(inverse)
print(equilv)
```
